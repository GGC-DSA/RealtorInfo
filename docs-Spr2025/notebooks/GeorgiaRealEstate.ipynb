## Project Description
For this project, we aim to use a real estate dataset, supplemented by other datasets, to help a client make an informed decision when looking to purchase property in Georgia.​

## Dataset
We are using the **Georgia subset** of the [USA Real Estate Dataset](https://www.kaggle.com/datasets/ahmedshahriarsakib/usa-real-estate-dataset) from Kaggle.

The dataset contains information on various properties across the U.S. We will filter and clean the data to focus only on properties in Georgia.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = real_estate = pd.read_csv("/content/sample_data/realtor-data.zip.csv")
```

The dataset contains real estate data for multiple states. Since the focus of this project is on Georgia, we filtered the rows where the `state` column matched "Georgia".
   

```python
# Check available states
print(df['state'].unique())

# Filter dataset for Georgia (GA)
df_ga = df[df['state'].str.strip().str.lower() == 'georgia']

# Check the first few rows
df_ga.head()
```

Columns like `latitude`, `longitude`, or other identifiers (`id`) were dropped as they were not directly useful for our analysis.
Removing unnecessary columns simplifies the dataset and reduces memory usage.


```python
print(df_ga.isnull().sum())
```

```python
df_ga_cleaned = df_ga.dropna()
```

```python
df_ga_cleaned = df_ga.fillna({
    'acre_lot': 0,
    'prev_sold_date': 'Unknown'
})
```

```python
df_ga_cleaned.to_csv("georgia_real_estate_cleaned.csv", index=False)
```
