# Imports and reading in dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Real_Estate.csv')
df
```

```python
GA_real_estate = df[df['state'] == "Georgia"]
GA_real_estate
```

```python
GA_real_estate.isna().sum()
```

```python
# Seeing as the primary purpose of the project is to find a location that the client finds suitable, we have opted to omit entries where
# locational data was left blank
GA_real_estate = GA_real_estate.dropna(subset = ['street', 'city', 'zip_code'])
GA_real_estate.isna().sum()
```

```python
GA_NA_bed_bath = GA_real_estate[GA_real_estate['bed'].isna() | GA_real_estate['bath'].isna()]
len(GA_NA_bed_bath)
```

```python
# Upon doing some research, there are various reasons why bed, bath, or both may be left blank. while the most common reason is that
# the listing is for land, rather than a house, there are other reasons, such as an incomplete listing or intentional omission of the
# count. As such, I did not want to simply remove rows where bed or bath hads null values, as this can still provide valuable information
# I instead opted to create a new column to mark these so that the data was not lost, but could be filtered out if desired.

GA_real_estate['Special_property'] = (GA_real_estate['bed'].isna()) | (GA_real_estate['bath'].isna())
GA_real_estate
```
